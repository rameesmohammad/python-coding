```python
n=int(input("Enter total number of monkeys:"))
k=int(input("Enter number of banana monkey can eat:"))
j=int(input("Enter number of peanut monkey can eat:"))
m=int(input("Enter total number of banana:"))
p=int(input("Enter total number of peanut:"))
if(k==0 or j==0):
  print("INVALID INPUT")
else:
  n=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    n=n-1
  print("No. of monkeys in tree:",n)

```
