```python
n=int(input("enter total no:of candies in the jar"))
k=int(input("enter number of minimum candies that must be inside JAR ever."))
C=1
R=n
while(C==1):
  M=int(input("Enter amout of candies needed:"))
  if(M>n):
    print("invalid")
  else:
    R=R-M
    if(R<=k):
      R=n
    print("SOLD=",M)
    print("Remaining=",R)
    C=int(input("DO you want to continue(1or0):"))
```

```python
n = int(input("Enter the no. of elements: "))
num = []
for i in range(n):
    a = int(input("Enter the element: "))
    num.append(a)
large = num[0]
for i in range(1, n):
    if num[i] > large:
        large = num[i]
print("largest element : ",large)
```
