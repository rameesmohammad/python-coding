```python
n=int(input("Enter total number of monkeys:"))
k=int(input("Enter number of banana monkey can eat:"))
j=int(input("Enter number of peanut monkey can eat:"))
m=int(input("Enter total number of banana:"))
p=int(input("Enter total number of peanut:"))
if(k==0 or j==0):
  print("INVALID INPUT")
else:
  n=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    n=n-1
  print("No. of monkeys in tree:",n)

```

```python
n=int(input("enter total no:of candies in the jar"))
k=int(input("enter number of minimum candies that must be inside JAR ever."))
C=1
R=n
while(C==1):
  M=int(input("Enter amout of candies needed:"))
  if(M>n):
    print("invalid")
  else:
    R=R-M
    if(R<=k):
      R=n
    print("SOLD=",M)
    print("Remaining=",R)
    C=int(input("DO you want to continue(1or0):"))
```
